#### 1. Create a class called `Book` that has the attributes title, author, pages, price. It also has the method get_price() which returns price, and set_price() to ensure price is numeric.

Then create two instances of this class:

b1:

* pages -> 383
* price -> 50

b2:
* title -> 'Building a Career in Data Science'
* author -> 'Robinson, Nolis'
* pages -> 322

```python
class Book:
  def __init__(self, title, author, pages, price):
    self.title = title
    self.author = author
    self.pages = pages
    # Use the set_price method to ensure price is numeric
    self.set_price(price)

  def get_price(self):
    return self.price

  def set_price(self, new_price):
    if isinstance(new_price, (int, float)):
      self.price = new_price
    else:
      try:
        self.price = float(new_price)
      except ValueError:
        print("Price must be a numeric value.")
        self.price = 0 # Default or handle error appropriately
```

#### 2. Create a `Phone` class that has the attributes model, id, name, brand and price. Create the getters and setters for each attribute. As well as the following methods: calculate_total which takes the price and calculates it by the tax argument.

```python
class Phone:
  def __init__(self, model, id_num, name, brand, price):
    self.model = model
    self.id = id_num # Renamed 'id' to 'id_num' to avoid conflict with built-in 'id'
    self.name = name
    self.brand = brand
    self.set_price(price) # Use setter for price validation

  # Getters
  def get_model(self):
    return self.model

  def get_id(self):
    return self.id

  def get_name(self):
    return self.name

  def get_brand(self):
    return self.brand

  def get_price(self):
    return self.price

  # Setters
  def set_model(self, new_model):
    self.model = new_model

  def set_id(self, new_id):
    self.id = new_id

  def set_name(self, new_name):
    self.name = new_name

  def set_brand(self, new_brand):
    self.brand = new_brand

  def set_price(self, new_price):
    if isinstance(new_price, (int, float)):
      self.price = new_price
    else:
      try:
        self.price = float(new_price)
      except ValueError:
        print("Price must be a numeric value.")
        self.price = 0 # Default to 0 or handle error appropriately

  # Method to calculate total price with tax
  def calculate_total(self, tax_rate):
    if not isinstance(tax_rate, (int, float)) or tax_rate < 0:
      print("Tax rate must be a non-negative numeric value.")
      return None
    return self.price * (1 + tax_rate)

```

```python
# Example usage (optional, for testing)
my_phone = Phone("iPhone 13 Pro", "XYZ123", "My Phone", "Apple", 999.99)
print(f"Phone brand: {my_phone.get_brand()}")
my_phone.set_price(1099.99)
print(f"New price: {my_phone.get_price()}")
print(f"Total with 8% tax: {my_phone.calculate_total(0.08)}")
```

#### 1. Write a code to convert a 1D array to a 2D array with 2 rows.

Original Array: [0 1 2 3 4 5]

Reshaped 2x3 Array:

[[0 1 2]

[3 4 5]]

```python
original_array = np.arange(6) # Creates a 1D array [0 1 2 3 4 5]
reshaped_array = original_array.reshape(2, 3) # Reshapes to 2 rows and 3 columns

print("Original Array:", original_array)
print("\nReshaped 2x3 Array:")
print(reshaped_array)
```

#### 4. Reverse a 1D NumPy array
Ex. arr = np.arange(10)

Output is [9 8 7 6 5 4 3 2 1 0]

*Hint! Use slicing with step = -1.*

```python
arr = np.arange(10)
reversed_arr = arr[::-1]

print("Original Array:", arr)
print("Reversed Array:", reversed_arr)
```

#### 6. Perform arithmetic operations on two NumPy arrays element-wise

a = np.array([1, 2, 3])

b = np.array([4, 5, 6])

*   Add two NumPy arrays element by element.
*   Multiply two NumPy arrays element by element.

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Add two NumPy arrays element by element
sum_array = a + b
print("Element-wise addition (a + b):", sum_array)

# Multiply two NumPy arrays element by element
product_array = a * b
print("Element-wise multiplication (a * b):", product_array)
```
